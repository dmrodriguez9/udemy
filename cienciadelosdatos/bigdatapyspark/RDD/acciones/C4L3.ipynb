Acciones en un RDD

*Son operaciones que activan los calculos

*Las acciones son las únicas operaciones que realmente activan los cálculos y ejecutan el DAG


Tipos de Acciones
- Driver

Se recompilan por conteos, por teclas etc y devuelve los datos al controlador
- Distribuidas

Se ejecutan en los nodos del cluster

```python
from pyspark.sql import SparkSession

spark= SparkSession.builder.getOrCreate()
sc= spark.sparkContext
```

Acciones: función reduce
aplica la reducción a todos los elementos del RDD y la envia al driver

```python
rdd= sc.parallelize([2,4,6,8])
rdd.reduce(lambda x,y: x+y)
```

```python
rdd1= sc.parallelize([1,2,3,4])
rdd1.reduce(lambda x,y: x*y)
```

Acciones: función count

*Cuenta el números de elementos en el RDD y lo envia al controlador

Internamente, El controlador le pide a cada ejecutor que cuente y luego suma los recuentos de todas las tareas juntas en el nivel de controlador


```python
rdd= sc.parallelize(['j', 'o', 's', 'e'])
rdd.count()
```

```python
rdd1= sc.parallelize([item for item in range(10)])
rdd1.count()
```

Acciones: función collect

*Recopila todos los elementos del RDD y lo envia al controlador

```python
rdd= sc.parallelize('Hola Apache Spark!'. split(' '))
rdd.collect()
```

```python
rdd1=sc.parallelize([(item, item **2) for item in range(20)])
rdd1.collect()
```

Acciones: funciones take, max y saveAsTextFile

Función Take

Devuelve los primeros elementos en el RDD

```python
rdd=sc.parallelize('La programación es bella'.split(' '))
rdd.take(2)
```

```python
rdd.take(4)
```

Función max

Devuelve el maximo de los números que encuentre en un RDD 

```python
rdd1=sc.parallelize([item/(item+1) for item in range(10)])
rdd1.max()
```

```python
rdd1.collect()
```

Función saveAsTextFile

*Son aquellas que se ejecutan en los nodos del clúster y el ejemplo es la función saveAsTextFile

```python
rdd=sc.parallelize('La programación es bella'.split(' '))
rdd.collect()
```

```python
rdd.saveAsTextFile('./acciones')
```

```python
rdd.coalesce(1).saveAsTextFile('./rdd1')
```
