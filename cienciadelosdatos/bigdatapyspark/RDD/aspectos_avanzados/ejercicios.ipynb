```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.getOrCreate()

sc=spark.sparkContext
```

1.- Cree un RDD importes a partir de los datos adjuntos a esta lección como recurso. Emplee acumuladores para obtener el total de ventas realizadas y el importe total de las ventas.

```python
rdd= sc.textFile('/content/rdd.txt')
```

```python
rdd_importes = rdd.map(int)
rdd_importes.collect()
```

```python
acumulador_total_ventas= sc.accumulator(0)
rdd_importes.foreach(lambda x: acumulador_total_ventas.add(1))
print(acumulador_total_ventas.value)
```

```python
acumulador_importe_total= sc.accumulator(0)
rdd_importes.foreach(lambda x: acumulador_importe_total.add(x))
print(acumulador_importe_total.value)
```

2.- Si se conoce que a cada venta hay que restarle un importe fijo igual a 10 pesos por temas de impuestos.

a. ¿Cómo restaría este impuesto de cada venta utilizando una variable broadcast para acelerar el proceso?

```python
impuesto_br_10=sc.broadcast(10)
```

b. Cree un RDD llamado ventas_sin_impuestos a partir de la propuesta del inciso a que contenga las ventas sin impuestos.

```python
rdd_ventas_sin_impuesto=rdd_importes.map(lambda x: x - impuesto_br_10.value)
rdd_ventas_sin_impuesto.collect()
```

c. Destruya la variable broadcast creada luego de emplearla para crear el RDD del inciso b.

```python
impuesto_br_10.destroy()
```

3.- Persista el RDD ventas_sin_impuestos en los siguientes niveles de persistencia.

```python
from pyspark.storagelevel import StorageLevel
```

a. Memoria.

```python
rdd_ventas_sin_impuesto.persist(StorageLevel.MEMORY_ONLY)
```

```python
rdd_ventas_sin_impuesto.unpersist()
```

b. Disco solamente

```python
rdd_ventas_sin_impuesto.persist(StorageLevel.DISK_ONLY)
```

```python
rdd_ventas_sin_impuesto.unpersist()
```

c. Memoria y disco.

```python
rdd_ventas_sin_impuesto.persist(StorageLevel.MEMORY_AND_DISK)
```

```python
rdd_ventas_sin_impuesto.unpersist()
```
