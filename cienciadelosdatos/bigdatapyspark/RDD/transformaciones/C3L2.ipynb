Transformaciones RDD

Los RDD son inmutables y cada operación crea un nuevo RDD

*Operaciones principales
- Transformaciones

  Dividir el elemento de entrada

  Filtrar elemenos
  
  Realizar cálculos de algún tipo

- Acciones

  

Categorías de las transformaciones

- Transformaciones generales

    ejm: map, filter, flatMap, groupByKey, sortByKey, combineByKey
- Transformaciones matemáticas o estadísticas

    ejm: sampleByKey, randomSplit
- Transformaciones de conjunto o relaciones

    ejm: cogroup, join, subtractByKey, fullOuterJoin, leftOuterJoin, rightOuterJoin
- Transformaciones basadas en estructura de datos

    ejm: partitionBy, repartition, zipwithIndex, coalesce

Transformaciones: funcion map

```python
from pyspark.sql import SparkSession

spark= SparkSession.builder.getOrCreate()
```

```python
sc= spark.sparkContext
```

```python
rdd= sc.parallelize([1,2,3,4,5])
```

```python
rdd_resta=rdd.map(lambda x:x-1)
rdd_resta.collect()
```

```python
rdd_par= rdd.map(lambda x:x%2==0)
rdd_par.collect()
```

```python
rdd_texto= sc.parallelize(['jose','juan','lucia'])
rdd_mayuscula= rdd_texto.map(lambda x:x.upper())
rdd_mayuscula.collect()
```

```python
rdd_hola= rdd_texto.map( lambda x: 'Hola' +" "+ x)
rdd_hola.collect()
```

Transformaciones: función flatMap (Particiones de entrada a particiones de salida, aplana)

```python
rdd= sc.parallelize([1,2,3,4,5])
rdd_cuadrado= rdd.map(lambda x:(x,x **2)) #** operador de potencia
rdd_cuadrado.collect()
```

```python
rdd_cuadrado_flat= rdd.flatMap(lambda x: (x, x**2))
rdd_cuadrado_flat.collect()
```

```python
rdd_texto= sc.parallelize(['jose','juan','juan'])
rdd_mayuscula=rdd_texto.flatMap(lambda x:(x,x.upper()))
rdd_mayuscula.collect()
```

Transformaciones: funcion filter

```python
rdd= sc.parallelize([1,2,3,4,5,6,7,8,9])
```

```python
rdd_par= rdd.filter(lambda x: x%2==0)
rdd_par.collect()
```

```python
rdd_impar= rdd.filter(lambda x:x%2!=0)
rdd_impar.collect()
```

```python
rdd_texto=sc.parallelize(['jose', 'joaquin', 'juan', 'lucia', 'karla', 'katia'])
rdd_k= rdd_texto.filter( lambda x: x.startswith('k'))
rdd_k.collect()
```

```python
rdd_filtro= rdd_texto.filter(lambda x:x.startswith('j') and x.find('u')==1) #1 es la posicion
rdd_filtro.collect()
```

Transformaciones: función coalesce

```python
rdd = sc.parallelize([1,2,3,4,5], 10)
rdd.getNumPartitions()
```

```python
rdd_5= rdd.coalesce(5)
rdd_5.getNumPartitions()
```

Transformaciones: función repartition (aumentar o disminuir particiones)

```python
rdd= sc.parallelize([1,2,3,4,5], 3)
rdd.getNumPartitions()
```

```python
rdd_7= rdd.repartition(7)
rdd_7.getNumPartitions()
```

Resumen : coalesce vs repartition
coalesce se usa solo para reducir el número de particiones. Esta es una versión optimizada de repatition donde el movimiento de los datos a través de las particiones es menor.

Importante: repartition y coalesce son operaciones muy costosas ya que mezclan los datos en muchas particiones, por lo tanto, intente minimizar el uso de estos tanto como sea posible.

Transformaciones: función reduceByKey

```python
rdd= sc.parallelize(
    [('casa',2),
     ('parque',1),
     ('que',5),
     ('casa',1),
     ('escuela', 2),
     ('casa',1),
     ('que',1)]
)
```

```python
rdd.collect()
```

```python
rdd_reducido = rdd.reduceByKey(lambda x,y: x+y)
rdd_reducido.collect()
```
