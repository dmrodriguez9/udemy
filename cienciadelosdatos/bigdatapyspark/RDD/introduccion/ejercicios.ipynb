Ejercicio 3

Cree un RDD que contenga los números primos que hay entre 1 y 20

```python
#range(inicio, fin)
rdd= sc.parallelize(range(1,21))
rdd_primos = rdd.filter( lambda x: x>1 and all (x%i !=0 for i in range(2,x)))
rdd_primos.collect()
```

Ejercicio 4

Cree un nuevo RDD a partir del RDD creado en el ejercicio anterior el cuál solo contenga los números primos mayores a 10.

```python
rdd_primos_may_10= rdd_primos.filter(lambda  x:x>10)
rdd_primos_may_10.collect()
```
