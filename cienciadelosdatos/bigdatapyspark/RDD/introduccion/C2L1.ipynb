RDD (Resilient Distributed Dataset)
Caracteristicas
- Dependencias
- Particiones
- Función de cálculo

Creando RDD de diferentes formas

```python
sc= spark.sparkContext
```

```python
#Creando un RDD Vacio
rdd_vacio= sc.emptyRDD()
```

```python
#Creando un RDD con parallelize
rdd_vacio3 = sc.parallelize([], 3)
rdd_vacio3.getNumPartitions()
```

```python
rdd = sc.parallelize([1,2,3,4,5])
rdd.collect()
```

- wholeTextFiles se utiliza cuando se necesita trabajar con el archivo completo como una sola unidad.

- textFile se utiliza cuando se necesita trabajar con cada línea del archivo como un registro independiente.

```python
#Creadno un RDD con un archivo de texto
rdd_texto = sc.textFile('rdd_source.txt')
rdd_texto.collect()
```

```python
rdd_texto_completo = sc.wholeTextFiles('rdd_source.txt')
rdd_texto_completo.collect()
```

```python
#map sirve para cambiar el valor de cada elemento
rdd_suma = rdd.map(lambda x: x +1)
rdd_suma.collect()
```

```python
#Creacion de un RDD a partir de un dataframe
df= spark.createDataFrame(
    [(1, 'Danny'),(2, 'Daniela')],
    ['id', 'nombre']
)
df.show()
```

```python
rdd_df= df.rdd
rdd_df.collect()
```
